```python
def convert (obj):
    L=[]
    for i in ast.literal_eval(obj):
       L.append(i['name'])
    return L
```

```python
def convert3 (obj):
    L=[]
    counter=0
    for i in ast.literal_eval(obj):
      if counter!=3:
           L.append(i['name'])
           counter+=1
      else:
          break
    return L
```

```python

import ast
import pandas as pd

def fetchdirector(obj):

    L = []

    if pd.isna(obj):
        return L

    for i in ast.literal_eval(obj):

        if i['job'] == 'Director':
            L.append(i['name'])
            break

    return L
```

```python
def stem(text):
    y=[]
    
    for i in text.split():
        y.append(ps.stem(i))
        
    return " ".join(y)
```

```python
def recommend_movie(movie):

    movie_index = df[df['title'] == movie].index[0]

    distances = similarity[movie_index]

    movies_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]

    for i in movies_list:
        print(df.iloc[i[0]].title)
```

```python
recommend_movie('Avatar')
```
